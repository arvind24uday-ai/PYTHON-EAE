1. Create a class "Human" that has 4 attributes: "ojos" (eyes, int default 2), "edad" (age, int), "altura" (height, float), and "nombre" (name, str). In addition, it must have two methods:

- "cumplir_edad": when calling this method, it should add 1 to the human's age attribute and return the updated age.
- "registro_nombre" which receives the parameter "nuevo_nombre": when calling this function with an input parameter, this method should change the human's name to "nuevo_nombre" and not return anything.

2. Create a human named "Cojoncio" who is twice your age (from the variable `lista_examen`) and is 10 centimeters shorter than your height (from the variable `lista_examen`). "Cojoncio" never liked his name. Therefore, use the corresponding method to change his name to "Goku". Also, today is his birthday, so use the corresponding method to increase his age. Show both the name and age before and after executing the methods.

3. Document the class in Google style.


```python
class Human:
    def __init__(self, nombre, edad, altura, ojos=2):
        self.nombre = nombre
        self.edad = edad
        self.altura = altura
        self.ojos = ojos

    def cumplir_edad(self):
        self.edad = self.edad + 1
        return self.edad

    def registro_nombre(self, nuevo_nombre):
        self.nombre = nuevo_nombre
```

```python
mi_edad = 29
mi_altura = 178.0 

cojoncio = Human(
    nombre="Cojoncio",
    edad=mi_edad * 2,       
    altura=mi_altura - 10   
)

print("Before changes:")
print("Name:", cojoncio.nombre)
print("Age:", cojoncio.edad)

cojoncio.registro_nombre("Goku")

cojoncio.cumplir_edad()

print("\nAfter changes:")
print("Name:", cojoncio.nombre)
print("Age:", cojoncio.edad)
```
