## Exercise 5
Given the following list, delete all values equal to 3
> ```ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]```

TIP: Do not try to delete elements from the list you are iterating. Make a copy with `ejer_5.copy()`.

````python
ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]

# Making a copy
for value in ejer_5.copy():
    if value == 3:
        ejer_5.remove(value)

print(ejer_5)
````

## Exercise 12
Convert the following list into a dictionary
> ```ejer_12 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]```

TIP: You can use two control variables in a `for` loop. One for each value of the tuples.

````python
ejer_12 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]

result_dict = {}

for key, value in ejer_12:
    if key not in result_dict:
        result_dict[key] = []
    result_dict[key].append(value)

print(result_dict)
````

## Exercise 13
1. Create an ascending sorted list with the dictionary keys
2. Create another list sorted in descending order with the values
3. Add a new key/value
4. Find the key with value 2 in the dictionary
5. Iterate the key and value of the dictionary with a single for loop
> ```ejer_13 = {4:78, 2:98, 8:234, 5:29}```

````python
ejer_13 = {4: 78, 2: 98, 8: 234, 5: 29}

# 1. Ascending sorted list with dictionary keys
sorted_keys = sorted(ejer_13.keys())

# 2. Descending sorted list with the values
sorted_values_desc = sorted(ejer_13.values(), reverse=True)

# 3. Add a new key/value
ejer_13[10] = 500

# 4. Find the key with value 2 (if it exists)
key_with_value_2 = None
for key, value in ejer_13.items():
    if value == 2:
        key_with_value_2 = key
        break

# 5. Iterate key and value with single for loop
for key, value in ejer_13.items():
    print(f"Key: {key}, Value: {value}")

print("Sorted keys:", sorted_keys)
print("Sorted values (desc):", sorted_values_desc)
print("Key with value 2:", key_with_value_2)
````

## Exercise 17
1. Create a set with three elements
2. Add a fourth
3. Delete the last element added
4. Delete element 10, if it is present. Use `discard()`

````python
# 1. Create a set with three elements
my_set = {1, 2, 3}

# 2. Add a fourth element
my_set.add(4)

print("After adding 4:", my_set)

# 3. Delete the last element added
# (Sets are unordered, so we remove the specific value we added)
my_set.remove(4)

print("After removing 4:", my_set)

# 4. Delete element 10 if present
my_set.discard(10)   # No error if 10 does not exist

print("Final set:", my_set)
````

Sets are unordered, so there is no concept of "last element".
That’s why we remove the value (4) instead of something like an index.
