## Exercise 2
Write a program to convert degrees to radians. You must use `input`. Remember that the conversion is done through

radians = degrees*(pi/180)

````python
import math
degrees = float(input("Enter degrees: "))
radians = degrees * (math.pi / 180)
print("Radians:", radians)
````

## Exercise 3
Write a program that calculates the area of a parallelogram (base x height). Also with `input`

````python
base = float(input("Enter the base of the parallelogram: "))
height = float(input("Enter the height of the parallelogram: "))
area = base * height
print("The area of the parallelogram is:", area)
````

## Exercise 5
Create a program where two user inputs are collected, and the program output is whether those inputs are equal or not

````python
user1 = input("Enter the first value: ")
user2 = input("Enter the second value: ")

if user1 == user2:
    print("The values are equal.")

if user1 != user2:
    print("The values are not equal.")
````

## Exercise 6
Same program as in 5, but this time there must be three inputs and two outputs. One output should tell us if all are equal, and the other if at least two inputs are equal

````python
value1 = input("Enter the first value: ")
value2 = input("Enter the second value: ")
value3 = input("Enter the third value: ")


if value1 == value2 == value3:
    print("All values are equal.")


if (value1 == value2) or (value1 == value3) or (value2 == value3):
    print("At least two values are equal.")
````

## Exercise 7
Create a program that takes two inputs. It must check if their sum is equal, greater than, or less than 10

````python
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

total = num1 + num2

print("Equal to 10:", total == 10)
print("Greater than 10:", total > 10)
print("Less than 10:", total < 10)
````

## Exercise 9
For this exercise we are going to practice [built-in functions](https://docs.python.org/3/library/functions.html).

1. Calculate the maximum of the list: [4, 6, 8, -1]
2. Sum all elements of the previous list
3. Round this float to 3 decimal digits: 63.451256965
4. Absolute value of: -74

````python
numbers = [4, 6, 8, -1]
maximum = max(numbers)
print("Maximum:", maximum)

total = sum(numbers)
print("Sum:", total)

value = 63.451256965
rounded_value = round(value, 3)
print("Rounded:", rounded_value)

absolute_value = abs(-74)
print("Absolute value:", absolute_value)
````

## Exercise 10
For the following string you are asked to print the following cases on screen

    "Early birds catch the worm"

1. Convert it all to uppercase
2. Convert it all to lowercase
3. Convert to uppercase only the initial of the words
4. Create a list by dividing it by its spaces
5. Replace commas `,` with semicolons `;`
6. Remove lowercase `a` letters

````python
text = "Early birds catch the worm"

print(text.upper())

print(text.lower())

print(text.title())

print(text.split(" "))

print(text.replace(",", ";"))

print(text.replace("a", ""))
````

## Exercise 11
1. Create a list with 3 numeric elements
2. Add a fourth element
3. Calculate the sum of all
4. Delete the second element of the list
5. Add another element at position 3 of the list
6. Create another list with 4 elements and concatenate it to the one you already had.

````python
numbers = [10, 20, 30]
print("Step 1:", numbers)

numbers.append(40)
print("Step 2:", numbers)

total = sum(numbers)
print("Step 3 - Sum:", total)

del numbers[1]
print("Step 4:", numbers)

numbers.insert(2, 25)
print("Step 5:", numbers)

other_list = [50, 60, 70, 80]
numbers = numbers + other_list
print("Step 6:", numbers)
````
