## Exercise 1
We are a company that has several appliance stores. We need a program to manage the stores, see the sales they have had, how many employees there are, etc... For this we are going to model the store in a class, which will have the following characteristics:
* Class name: "Store"
* Common attributes:
    * Type: "Appliances"
    * Open: True
* Attributes specific to each store:
    * Name - String
    * Address - String
    * Number of employees - int
    * Sales last 3 months - List of 3 numbers
* Method to calculate sales from all months, returning a number with all sales.
* Method that calculates the average sales of the last months, per employee
* Method that returns in a string the name of the store, along with its address.
* Method that gets the sales from the last month.
* Method to give a sales projection in case X money has been invested in marketing for that store. Where X is the input parameter. If X is less than 1000, sales for the last 3 months would have been (1.2 x sales), if greater than or equal to 1000, sales would have been (1.5 x sales). The method rewrites the sales attribute for the last 3 months, and also returns the sales with the increase.

It is requested:
1. Implement the "Store" class
2. Create three stores with invented data
3. Check in at least one of them, everything implemented in the store class (its attributes, average sales, sales/employee...)
3. Calculate the sales from the last month of all stores. For this use the `for` loop
4. Print on screen the names of the stores whose address contains the string "Avenue"
5. I'm sure you've already done this section... Document the class :)

```python
from __future__ import annotations
from dataclasses import dataclass, field
from typing import List


@dataclass
class Store:
    """
    Represents an appliance store owned by the company.

    Common attributes (same for all stores):
        - type: str = "Appliances"
        - open: bool = True

    Store-specific attributes:
        - name: str
        - address: str
        - num_employees: int
        - sales_last_3_months: list[float] of length 3 (oldest -> newest)

    Methods:
        - total_sales(): sum of the last 3 months of sales
        - average_sales_per_employee(): average monthly sales (last 3 months) per employee
        - store_label(): "Name - Address"
        - last_month_sales(): sales from the most recent month
        - project_sales(investment): applies a multiplier (1.2 or 1.5) to all 3 months,
          REWRITES sales_last_3_months, and returns the updated sales list
    """

    # Common attributes
    type: str = field(default="Appliances", init=False)
    open: bool = field(default=True, init=False)

    # Store-specific attributes
    name: str
    address: str
    num_employees: int
    sales_last_3_months: List[float]

    def __post_init__(self) -> None:
        if not isinstance(self.name, str) or not self.name.strip():
            raise ValueError("name must be a non-empty string")

        if not isinstance(self.address, str) or not self.address.strip():
            raise ValueError("address must be a non-empty string")

        if not isinstance(self.num_employees, int) or self.num_employees <= 0:
            raise ValueError("num_employees must be a positive integer")

        if (
            not isinstance(self.sales_last_3_months, list)
            or len(self.sales_last_3_months) != 3
        ):
            raise ValueError("sales_last_3_months must be a list of exactly 3 numbers")

        for s in self.sales_last_3_months:
            if not isinstance(s, (int, float)) or s < 0:
                raise ValueError("each sales value must be a non-negative number")

        # Normalize to float for consistent calculations
        self.sales_last_3_months = [float(s) for s in self.sales_last_3_months]

    def total_sales(self) -> float:
        """Return the sum of sales for the last 3 months."""
        return sum(self.sales_last_3_months)

    def average_sales_per_employee(self) -> float:
        """
        Return the average sales of the last 3 months, per employee.
        (Average monthly sales across last 3 months) / num_employees
        """
        avg_monthly = self.total_sales() / 3
        return avg_monthly / self.num_employees

    def store_label(self) -> str:
        """Return a string with the name of the store and its address."""
        return f"{self.name} - {self.address}"

    def last_month_sales(self) -> float:
        """Return the sales from the most recent month (the 3rd value)."""
        return self.sales_last_3_months[-1]

    def project_sales(self, investment: float) -> List[float]:
        """
        Apply a sales multiplier based on marketing investment, rewrite sales_last_3_months,
        and return the updated sales list.

        Rules:
            - if investment < 1000: multiplier = 1.2
            - else: multiplier = 1.5
        """
        if not isinstance(investment, (int, float)) or investment < 0:
            raise ValueError("investment must be a non-negative number")

        multiplier = 1.2 if investment < 1000 else 1.5
        self.sales_last_3_months = [s * multiplier for s in self.sales_last_3_months]
        return self.sales_last_3_months
```

```python
# 2) Create three stores with invented data
# ------------------------------------------------------------
stores = [
    Store(
        name="Northside Appliances",
        address="123 Maple Avenue, Springfield",
        num_employees=8,
        sales_last_3_months=[52000, 61000, 57500],
    ),
    Store(
        name="Downtown Home Tech",
        address="45 River Street, Springfield",
        num_employees=12,
        sales_last_3_months=[83000, 79000, 84500],
    ),
    Store(
        name="Lakeside Electronics & Appliances",
        address="900 Lakeview Avenue, Shelbyville",
        num_employees=6,
        sales_last_3_months=[41000, 43000, 45500],
    ),
]
```

```python
# 3) Check in at least one store everything implemented
# ------------------------------------------------------------
test_store = stores[0]
print("=== Store check (example: first store) ===")
print("Common attributes:", test_store.type, test_store.open)
print("Label:", test_store.store_label())
print("Total sales (3 months):", test_store.total_sales())
print("Last month sales:", test_store.last_month_sales())
print("Avg sales per employee:", test_store.average_sales_per_employee())

print("\nApplying projection with investment = 800 (should be 1.2x)")
updated_sales = test_store.project_sales(800)
print("Updated sales_last_3_months:", updated_sales)
print("New total sales:", test_store.total_sales())
```

```python
# 4) Calculate sales from the last month of all stores (use for loop)
# ------------------------------------------------------------
total_last_month_all_stores = 0.0
for st in stores:
    total_last_month_all_stores += st.last_month_sales()

print("\n=== Last month sales (all stores) ===")
print("Total last-month sales across all stores:", total_last_month_all_stores)
```

```python
# 5) Print names of stores whose address contains "Avenue"
# ------------------------------------------------------------
print('\n=== Stores with "Avenue" in address ===')
for st in stores:
    if "Avenue" in st.address:
        print(st.name)
```

## Exercise 2
In this exercise we are going to implement a *Dog* class in Python. The class has the following characteristics:
* Things we know for sure a dog has
    * It has 4 legs
    * 2 ears
    * 2 eyes
    * A speed of 0. By default, the dog is stopped
* When initialized:
    * The dog will be of a certain breed
    * By default it will have "Brown" fur, unless otherwise stated.
    * By default it will have no owner, unless otherwise stated.
    
* It will also have a method called walk, which has an input argument (speed_increase). This value will be added to the speed the dog already had.
* It needs another method (stop), where we will set the speed to 0.
* Another method called "bark", which will have an input argument, and the output will be the following string: "WOOF!" + the input argument.


It is requested:
* Implement the *Dog* class
* Create an object of type *Dog*, without an owner
* Check that all its attributes are correct
* Test that it walks, and check its speed
* Stop it
* Document the *Dog* class

```python
class Dog:
   

    def __init__(self, breed, fur_color="Brown", owner=None):
        """Initialize a new Dog object."""
        self.breed = breed
        self.fur_color = fur_color
        self.owner = owner
        self.legs = 4
        self.ears = 2
        self.eyes = 2
        self.speed = 0

    def walk(self, speed_increase):
        """Increase the dog's speed."""
        self.speed += speed_increase

    def stop(self):
        """Stop the dog (set speed to 0)."""
        self.speed = 0

    def bark(self, message):
        """Return a barking message."""
        return "WOOF! " + message
```

```python
my_dog = Dog(breed="Labrador")
```

```python
print("Breed:", my_dog.breed)
print("Fur Color:", my_dog.fur_color)
print("Owner:", my_dog.owner)
print("Legs:", my_dog.legs)
print("Ears:", my_dog.ears)
print("Eyes:", my_dog.eyes)
print("Speed:", my_dog.speed)
```

```python
my_dog.walk(5)
print("Speed after walking:", my_dog.speed)
```

```python
my_dog.stop()
print("Speed after stopping:", my_dog.speed)
```

```python
print(my_dog.bark("I'm happy!"))
```
