## Exercise 1
Write a function that converts numbers from 1 to 7 into names of the days of the week. The function will consist of a single numeric argument and a string type output

````python
def number_to_day(number):
    days = {
        1: "Monday",
        2: "Tuesday",
        3: "Wednesday",
        4: "Thursday",
        5: "Friday",
        6: "Saturday",
        7: "Sunday"
    }
    
    return days.get(number, "Invalid number")
print(number_to_day(1)) 
print(number_to_day(4))  
print(number_to_day(8))  
````

## Exercise 2
In exercise 8 of loops, we created an inverted pyramid, whose number of floors was determined by user input. Create a function that replicates the behavior of the pyramid, and uses a single input parameter of the function to determine the number of rows of the pyramid, that is, removes the input statement.

````python
def pyramid(rows):

    while rows > 0:
     for i in range(rows, 0, -1):
        print(i, end="")
     print() 
     rows = rows - 1   

    return "pyramid completed"
````

````python
pyramid(5)
````

## Exercise 3
Write a function that compares two numbers. The function has two arguments and there are three possible outputs: they are equal, the first is greater than the second, or the second is greater than the first

````python
def compare_numbers(num1, num2):

    num1 = float(num1)
    num2 = float(num2)
    if num1 == num2:
        return "They are equal"
    elif num1 > num2:
        return str(num1) + "The first is greater than the second" + str(num2)
    else:
        return str(num2) + "The second is greater than the first" + str(num1)

print(compare_numbers(5, 3))
print(compare_numbers(2, 7))
print(compare_numbers(4, 4))
````

## Exercise 4
Write a function that is a letter counter. In the first argument you have to introduce a text, and the second that is the letter to count. The function must return an integer with the number of times that letter appears, both uppercase and lowercase

````python
def letter_counter(text, letter):
    text = text.lower()
    return text.count(letter.lower())
   
  
        
````

````python
letter_counter("Hello World", "O") 
````

## Exercise 5
Write a function that has a single argument, a string. The output of the function must be a dictionary with the count of all the letters of that string.

````python
def count_letters(text):
    result = {}
    
    for char in text:
        if char in result:
            result[char] += 1
        else:
            result[char] = 1
            
    return result
print(count_letters("cat is cute"))
````

## Exercise 6
Write a function that adds or removes elements in a list. The function needs the following arguments:
* list: the list where elements will be added or removed
* command: "add" or "remove"
* element: By default it is None.

````python
def modify_list(my_list, command, element=None):
    if command == "add" and element is not None:
        my_list.append(element)

    elif command == "remove" and element in my_list:
        my_list.remove(element)
        
    else:
        print("Invalid command or element not found")


numbers = [1, 2, 3]

modify_list(numbers, "add", 7)
print(numbers)

modify_list(numbers, "remove", 1)
print(numbers)
````

## Exercise 7
Create a function that receives an arbitrary number of words, and returns a complete sentence, separating the words with spaces.

````python
def create_sentence(*words):
    return " ".join(words)

print(create_sentence("Python", "is", "very", "powerful"))
print(create_sentence("I", "love", "coding"))
````

## Exercise 8
Write a program that obtains the nth number of the [Fibonacci sequence](https://en.wikipedia.org/wiki/Fibonacci_number). You must create a recursive function with a single argument.

```Python
fibonacci(2) = 1
fibonacci(4) = 3
fibonacci(5) = 5
fibonacci(8) = 21
```

````python
def fibonacci(n):
    if n <= 0:
        return "Invalid input"
    elif n == 1:
        return 0
    elif n == 2:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)
print("fibonacci(2) =", fibonacci(2))
print("fibonacci(4) =", fibonacci(4))
print("fibonacci(5) =", fibonacci(5))
print("fibonacci(8) =", fibonacci(8))
````

## Exercise 9
Define in a single cell the following functions:
* Function that calculates the area of a square
* Function that calculates the area of a triangle
* Function that calculates the area of a circle

In another cell, calculate the area of:
* Two circles with radius 10 + a triangle with base 3 and height 7
* A square with side = 10 + 3 circles (one with radius = 4 and the other two with radius = 6) + 5 triangles with base = 2 + height = 4

````python
import math

def area_square(side):
    return side ** 2

def area_triangle(base, height):
    return (base * height) / 2

def area_circle(radius):
    return math.pi * (radius ** 2)
````

````python
# 1) Two circles with radius 10 + a triangle with base 3 and height 7
# 2) A square with side 10 + 3 circles (one radius 4 and two radius 6)
#    + 5 triangles with base 2 and height 4

result_1 = 2 * area_circle(10) + area_triangle(3, 7)
print("Result 1 =", result_1)

result_2 = area_square(10) + area_circle(4) + 2 * area_circle(6) + 5 * area_triangle(2, 4)
print("Result 2 =", result_2)
````
